# Үй жұмысы: мәліметтер құрылымдары және файлдармен жұмыс

## Интернет-дүкен тапсырыстарын талдау

Сіз шағын интернет-дүкеннің деректер талдаушысысыз. Дүкен басшылығы қазан айындағы 10 тапсырыс жазылған `orders.txt` файлын берді. Үй жұмысы бір үлкен есептен тұрады: файлды оқу, деректерді тізімдер, кортеждер, жиындар және сөздіктер арқылы талдау, нәтижелерді жаңа файлдарға сақтау.

Тапсырмалар үш деңгейге бөлінген және біртіндеп күрделене түседі:

- **А деңгейі (1–10):** кодтың негізгі бөлігі дайын, тек бірнеше жолды толықтыру керек;
- **В деңгейі (11–20):** кодтың құрылымы (циклдер, шарттар) берілген, ішкі логиканы өзіңіз жазасыз;
- Қосымша орындауға **С деңгейі (21–30):** тек функцияның атауы мен сипаттамасы берілген, шешімді толық өзіңіз құрасыз.

**Жұмыс тәртібі:**

- тапсырмаларды ретімен орындаңыз: әр тапсырма алдыңғы тапсырмаларда құрылған айнымалыларды қолданады;
- код ұяшығындағы `# Шешіміңізді осында жазыңыз` деген жерге өз кодыңызды жазыңыз;
- `...` белгісі тұрған жерге тиісті өрнекті қойыңыз;
- дайын берілген код жолдарын өзгертпеңіз, қажет болса жаңа жолдар қосыңыз;
- нәтижеңізді "Өзін-өзі тексеру" бөліміндегі мәнмен салыстырыңыз.

## Деректер файлының сипаттамасы

`orders.txt` — мәтіндік файл, кодтауы UTF-8. Бірінші жолда баған атаулары, қалған 10 жолдың әрқайсысында бір тапсырыс жазылған. Мәндер нүктелі үтір `;` арқылы бөлінген.

| Баған | Мағынасы | Мысал |
|---|---|---|
| `order_id` | тапсырыс нөмірі | `O101` |
| `date` | тапсырыс күні | `2026-10-01` |
| `customer` | клиенттің аты | `Aigerim` |
| `city` | жеткізу қаласы | `Almaty` |
| `category` | тауар санаты | `Electronics` |
| `product` | тауар атауы | `Mouse` |
| `price` | бір дананың бағасы, теңге | `6500` |
| `quantity` | дана саны | `2` |

Тапсырыс сомасы бағаны дана санына көбейту арқылы есептеледі: `price * quantity`.

## А деңгейі

**Мақсаты:** файлды оқуды, жолды бөліктерге бөлуді және төрт құрылымның әрқайсысын қарапайым түрде қолдануды меңгеру.

Бұл деңгейде кодтың басым бөлігі дайын. Сіздің міндетіңіз — `...` және `# Шешіміңізді осында жазыңыз` тұрған жерлерді толықтырып, әр жолдың не істейтінін түсіну.

### Тапсырма 1. Файлды оқу

`with` операторы арқылы `orders.txt` файлын ашып, барлық жолды `lines` тізіміне оқыңыз. Әр жолдың соңындағы `\n` таңбасын `strip()` арқылы алып тастаңыз да, барлық жолды экранға шығарыңыз.

**Өзін-өзі тексеру:** `lines` тізімінде 11 жол бар.

```python
with open("orders.txt", encoding="utf-8") as f:
    lines = []
    for line in f:
        lines.append(line.strip())

for line in lines:
    print(line)
```

### Тапсырма 2. Тақырып жолын бөлу

`lines` тізімінің бірінші элементі — баған атаулары. Оны `split(";")` арқылы `header` тізіміне айналдырыңыз. Қалған жолдарды кесінді арқылы `data_lines` тізіміне сақтаңыз. Баған атауларын реттік нөмірімен шығарыңыз.

**Өзін-өзі тексеру:** `header` тізімінде 8 атау, `data_lines` тізімінде 10 жол бар.

```python
header =lines[0].split(";")
data_lines =lines[1:]

for number, column in enumerate(header, start=1):
    print(number, column)

print("Тапсырыстар саны:", len(data_lines))
```

### Тапсырма 3. Бір жолды бөліп алу

`data_lines` тізімінің бірінші жолын `;` бойынша бөліп, `fields` тізіміне сақтаңыз. Клиенттің атын, тауар атауын, бағасын және санын индекстер арқылы алыңыз. Баға мен санды бүтін санға түрлендіріп, тапсырыс сомасын есептеңіз.

**Өзін-өзі тексеру:** `Aigerim: Mouse, 2 дана, сомасы 13000 теңге`.

```python
fields = data_lines[0].split(";")
print(fields)

customer = fields[2]
product = fields[5]
price = int(fields[6])
quantity = int(fields[7] )

amount = price*quantity
print(f"{customer}: {product}, {quantity} дана, сомасы {amount} теңге")
```

### Тапсырма 4. Барлық тапсырыс: тізімдер тізімі

`data_lines` тізімінің әр жолын бөліп, баға мен санды бүтін санға түрлендіріңіз және нәтижені `orders` тізіміне қосыңыз. Нәтижесінде `orders` тізімінің әр элементі бір тапсырыстың 8 өрісінен тұратын тізім болады.

```python
orders = []

for line in data_lines:
    fields = line.split(";")
    fields[6] = int(fields[6])
    # Шешіміңізді осында жазыңыз: quantity өрісін де бүтін санға түрлендіріңіз
    fields[7]=int(fields[7])
    orders.append(fields)

print("Бірінші тапсырыс:", orders[0])
print("Соңғы тапсырыс:", orders[-1])
```

### Тапсырма 5. Клиенттер тізімі

`orders` тізімінен барлық клиенттің атын (қайталанулармен бірге) `customers` тізіміне жинаңыз. Тізімнің ұзындығын, бірінші және соңғы элементін шығарыңыз. `count()` әдісі арқылы `Aigerim` неше рет тапсырыс бергенін анықтаңыз.

**Өзін-өзі тексеру:** `Aigerim` 3 рет тапсырыс берген.

```python
customers = []

for order in orders:
    customers.append(order[2])
    ...

print(customers)
print("Ұзындығы:", len(customers))
print("Бірінші:", customers[0], "соңғы:", customers[-1])   # Шешіміңізді осында жазыңыз
print("Aigerim:", customers.count("Aigerim"))                  # Шешіміңізді осында жазыңыз
```

### Тапсырма 6. Тапсырыс сомалары

Әр тапсырыстың сомасын `amounts` тізіміне жинаңыз. `sum()`, `max()`, `min()` және `len()` функциялары арқылы жалпы түсімді, ең үлкен, ең кіші және орташа соманы шығарыңыз.

**Өзін-өзі тексеру:** жалпы түсім 141100, ең үлкен сома 25000, ең кіші сома 7000.

```python
amounts = []

for order in orders:
    amount = order[6] * order[7]
    amounts.append(amount)
    ...

print(amounts)
print("Жалпы түсім:", sum(amounts))                     # Шешіміңізді осында жазыңыз
print("Ең үлкен сома:", max(amounts))                   # Шешіміңізді осында жазыңыз
print("Ең кіші сома:", min(amounts))                    # Шешіміңізді осында жазыңыз
print("Орташа сома:", sum(amounts)/len(amounts))                     # Шешіміңізді осында жазыңыз
```

### Тапсырма 7. Тауарлар тізімін сұрыптау

Барлық тауар атауларын `products` тізіміне жинап, оны `sort()` арқылы әліпби бойынша сұрыптаңыз. Тізімге `"Laptop"` тауарын `append()` арқылы қосып, содан кейін оны `remove()` арқылы өшіріңіз. Әр қадамнан кейін тізімді шығарыңыз.

```python
products = []
for order in orders:
    products.append(order[5])

# Шешіміңізді осында жазыңыз: тізімді сұрыптаңыз
products.sort()
print("Сұрыпталған:", products)

# Шешіміңізді осында жазыңыз: "Laptop" тауарын қосыңыз
products.append("Laptop")
print("Қосылғаннан кейін:", products)

# Шешіміңізді осында жазыңыз: "Laptop" тауарын өшіріңіз
products.remove("Laptop")
print("Өшірілгеннен кейін:", products)
```

### Тапсырма 8. Тапсырыс кортеж ретінде

Бірінші тапсырысты `tuple()` арқылы кортежге айналдырыңыз. Кортежді бөліп алу (unpacking) арқылы оның 8 өрісін сегіз айнымалыға меншіктеңіз. Кортеждің элементін өзгертуге тырысып, пайда болған қатені `try-except` арқылы ұстаңыз.

```python
first = tuple(orders[0])
print(first)

order_id, date, customer, city, category, product, price, quantity = first
print(order_id, date, customer)    # Шешіміңізді осында жазыңыз: тапсырыс нөмірін, күнін және клиентін шығарыңыз

try:
    first[6] = 1000
except TypeError as error:
    print("Қате:", error)
```

### Тапсырма 9. Бірегей клиенттер мен қалалар

`customers` тізімінен бірегей клиенттер жиынын жасаңыз. `orders` тізімі бойынша цикл арқылы бірегей қалалар жиынын толтырыңыз. Әр жиынның элементтер санын шығарыңыз.

**Өзін-өзі тексеру:** 5 бірегей клиент, 3 қала.

```python
unique_customers = set(customers)
print(unique_customers, len(unique_customers))

cities = set()
for order in orders:
    # Шешіміңізді осында жазыңыз
    cities.add(order[3])

print(cities, len(cities))
```

### Тапсырма 10. Тапсырыс сөздік ретінде

`zip()` функциясы арқылы `header` тізімі мен бірінші тапсырыстың өрістерін жұптап, `dict()` арқылы сөздікке айналдырыңыз. Сөздіктен мәндерді кілттер арқылы алып, тапсырыс туралы сөйлем шығарыңыз.

**Өзін-өзі тексеру:** `Aigerim Almaty қаласына Mouse тапсырыс берді`.

```python
first_dict = dict(zip(header, orders[0]))
print(first_dict)

customer = first_dict["customer"]
product = first_dict["product"]     # Шешіміңізді осында жазыңыз
city = first_dict["city"]        # Шешіміңізді осында жазыңыз

print(f"{customer} {city} қаласына {product} тапсырыс берді")
```

## В деңгейі

**Мақсаты:** тізімдік өрнектерді, сұрыптауды, жиын амалдарын және сөздікті санағыш ретінде қолдануды үйрену.

Бұл деңгейде кодтың құрылымы ғана берілген. Циклдің немесе шарттың ішіндегі логиканы, сондай-ақ кейбір өрнектерді өзіңіз жазасыз.

### Тапсырма 11. Сөздіктер тізімі

Әр тапсырысты 10-тапсырмадағыдай сөздікке айналдырып, `orders_dicts` тізіміне жинаңыз. Әр сөздікке тапсырыс сомасын сақтайтын `"amount"` кілтін қосыңыз. Бұдан былай негізінен осы тізіммен жұмыс істейміз.

```python
orders_dicts = []

for order in orders:
    # Шешіміңізді осында жазыңыз
    d=dict(zip(header,order))
    d["amount"]=order[6]*order[7]
    orders_dicts.append(d)

print(orders_dicts[0])
print(len(orders_dicts))
```

### Тапсырма 12. Тапсырыстарды сүзу

Тізімдік өрнек арқылы екі тізім құрыңыз: сомасы 10 000 теңгеден асатын тапсырыстардың нөмірлері және `Almaty` қаласына жеткізілген тапсырыстардың нөмірлері.

**Өзін-өзі тексеру:** 10 000-нан асатын 6 тапсырыс, Almaty қаласына 4 тапсырыс.

```python
big_orders = [order["order_id"] for order in orders_dicts if order["amount"]>10000]       # Шешіміңізді осында жазыңыз
almaty_orders = [order["order_id"] for order in orders_dicts if order["city"]=="Almaty"]    # Шешіміңізді осында жазыңыз

print("10 000-нан асатын:", big_orders)
print("Almaty:", almaty_orders)
```

### Тапсырма 13. Ең ірі 3 тапсырыс

`sorted()` функциясы мен `key` параметрін қолданып, `orders_dicts` тізімін сома бойынша кему ретімен сұрыптаңыз. Кесінді арқылы ең ірі 3 тапсырысты алып, кесте түрінде шығарыңыз.

**Өзін-өзі тексеру:** ең ірі тапсырыстар: O105, O110, O108.

```python
sorted_orders = sorted(orders_dicts, key=lambda o: o["amount"], reverse=True)   # Шешіміңізді осында жазыңыз
top3 = sorted_orders[:3]                                                     # Шешіміңізді осында жазыңыз

for o in top3:
    print(f"{o['order_id']}  {o['customer']:<8} {o['product']:<12} {o['amount']:>7}")
```

### Тапсырма 14. Кортеждер тізімі және zip

`zip()` арқылы `customers` және `amounts` тізімдерін `(клиент, сома)` кортеждерінің тізіміне біріктіріңіз. Цикл ішінде әр кортежді бөліп алып, ең ірі соманы және оның иесін табыңыз. `max()` функциясын қолданбаңыз.

**Өзін-өзі тексеру:** Arman, 25000.

```python
pairs = list(zip(customers, amounts))
print(pairs)

best_customer = ""
best_amount = 0

for customer, amount in pairs:
    # Шешіміңізді осында жазыңыз
    if amount>best_amount:
      best_amount=amount
      best_customer=customer

print("Ең ірі тапсырыс:", best_customer, best_amount)
```

### Тапсырма 15. Санаттар бойынша клиенттер

`Electronics` санатынан сатып алған клиенттер жиынын және `Books` санатынан сатып алған клиенттер жиынын құрыңыз. Жиын амалдары арқылы екі санаттан да сатып алғандарды, тек `Electronics` сатып алғандарды және кем дегенде бір санаттан сатып алғандарды табыңыз.

**Өзін-өзі тексеру:** екі санаттан да тек Aigerim сатып алған.

```python
electronics_buyers = set()
books_buyers = set()

for o in orders_dicts:
    if o["category"]=="Electronics":
        electronics_buyers.add(o["customer"])
    elif o["category"]=="Books":
        books_buyers.add(o["customer"])
    ...

print("Екеуінен де:",electronics_buyers.intersection(books_buyers) )            # Шешіміңізді осында жазыңыз
print("Тек Electronics:", electronics_buyers-books_buyers)        # Шешіміңізді осында жазыңыз
print("Кем дегенде біреуінен:", electronics_buyers.union(books_buyers))  # Шешіміңізді осында жазыңыз
```

### Тапсырма 16. Каталогпен салыстыру

Дүкен каталогы берілген. Сатылған тауарлардың жиынын құрып, каталогтағы, бірақ сатылмаған тауарларды табыңыз. `issubset()` әдісі арқылы барлық сатылған тауардың каталогта бар екенін тексеріңіз.

**Өзін-өзі тексеру:** сатылмаған тауарлар: Dictionary, Laptop.

```python
catalog = {"Mouse", "Keyboard", "Headphones", "Laptop", "Python Book",
           "Novel", "Dictionary", "Notebook", "Pen", "Backpack"}

sold = {o["product"] for o in orders_dicts}        # Шешіміңізді осында жазыңыз
not_sold = catalog-sold        # Шешіміңізді осында жазыңыз

print("Сатылған тауарлар:", sorted(sold))
print("Сатылмаған тауарлар:", sorted(not_sold))
print("Барлығы каталогта бар ма:", sold.issubset(catalog))   # Шешіміңізді осында жазыңыз
```

### Тапсырма 17. Клиенттер санағышы

### Тапсырма 21. Файлды оқитын функция

Файл жолын қабылдап, сөздіктер тізімін қайтаратын `load_orders(path)` функциясын жазыңыз. Әр сөздікте 8 баған және `"amount"` кілті болуы керек, ал `price` пен `quantity` бүтін сан болуы тиіс. Функцияны шақырып, нәтижені `orders_dicts` тізімімен салыстырыңыз.

**Өзін-өзі тексеру:** нәтиже `orders_dicts` тізімімен бірдей (`True`).

```python
def load_orders(path):
    """Файлды оқып, тапсырыстарды сөздіктер тізімі ретінде қайтарады."""
    # Шешіміңізді осында жазыңыз
    with open(path, "r", encoding="utf-8") as f:
      header=lines[0].split(";")
      data_lines=lines[1:]
      orders_dicts=[]
      for line in data_lines:
        fields=line.split(";")
        fields[6]=int(fields[6])
        fields[7]=int(fields[7])
        d=dict(zip(header,fields))
        d["amount"]=fields[6]*fields[7]
        orders_dicts.append(d)
    return orders_dicts


loaded = load_orders("orders.txt")
print(len(loaded))
print(loaded == orders_dicts)
```

### Тапсырма 22. Қалалар бойынша кірістірілген сөздік

Тапсырыстар тізімін қабылдап, кілті қала атауы, мәні `{"orders": тапсырыс саны, "revenue": түсім, "customers": клиенттер жиыны}` болатын сөздік қайтаратын `city_stats(orders)` функциясын жазыңыз. Нәтижені кесте түрінде шығарыңыз.

**Өзін-өзі тексеру:** Almaty: 4 тапсырыс, 47100 теңге.

```python
def city_stats(orders):
    """Әр қала бойынша тапсырыс санын, түсімді және клиенттер жиынын қайтарады."""
    # Шешіміңізді осында жазыңыз
    stats={}
    for o in orders:
      city=o["city"]
      if city not in stats:
        stats[city]={"orders":0, "revenue":0, "customers":set()}
      stats[city]["orders"]+=1
      stats[city]["revenue"]+=o["amount"]
      stats[city]["customers"].add(o["customer"])
    return stats;


stats = city_stats(orders_dicts)
for city, info in stats.items():
    print(f"{city:<10} {info['orders']:>2}  {info['revenue']:>7}  {sorted(info['customers'])}")
```

### Тапсырма 23. Клиенттер есебі

Тапсырыстар тізімін қабылдап, әр клиент үшін `{"orders": саны, "total": жалпы шығыны, "average": орташа чегі, "cities": қалалар жиыны}` сөздігін қайтаратын `customer_stats(orders)` функциясын жазыңыз. Нәтижеден бірнеше қаладан тапсырыс берген клиентті және ең көп ақша жұмсаған клиентті табыңыз.

**Өзін-өзі тексеру:** Dana екі қаладан тапсырыс берген, ең көп жұмсаған клиент Arman (44500 теңге).

```python
def customer_stats(orders):
    """Әр клиенттің тапсырыстары туралы жиынтық ақпаратты қайтарады."""
    # Шешіміңізді осында жазыңыз
    stats={}
    for o in orders:
      c=o["customer"]
      if c not in stats:
        stats[c]={"orders":0, "total": 0, "average":0, "cities":set()}
      stats[c]["orders"]+=1
      stats[c]["total"]+=o["amount"]
      stats[c]["cities"].add(o["city"])
    for c in stats:
      stats[c]["average"]=stats[c]["total"]/stats[c]["orders"]
    return stats

cstats = customer_stats(orders_dicts)
# Шешіміңізді осында жазыңыз: бірнеше қаладан тапсырыс берген клиентті және ең көп жұмсаған клиентті шығарыңыз
multi_city=max(cstats, key=lambda c: len(cstats[c]["cities"]))
print("Бірнеше қаладан тапсырыс берген: ", multi_city)
best_customer=max(cstats, key=lambda c: cstats[c]["total"])
print("Ең көп жұмсаған клиент: ", best_customer)
```

### Тапсырма 25. Қатеге төзімді оқу

Енді 21-тапсырмадағы `load_orders()` функциясы `ValueError` қатесімен тоқтайды. Жаңа `safe_load(path)` функциясын жазыңыз: ол түрлендіру кезінде қате шыққан жолдарды өткізіп жіберіп, `(тапсырыстар_тізімі, қате_жолдар_тізімі)` кортежін қайтаруы керек. Қате жолдар тізіміне жолдың файлдағы нөмірі мен мәтіні сақталсын. Нәтижені бөліп алып, екі тізімнің ұзындығын шығарыңыз.

**Өзін-өзі тексеру:** 12 дұрыс тапсырыс, 1 қате жол (файлдағы 14-жол). Жалпы түсім енді 205500 теңге.

```python
def safe_load(path):
    """Қате жолдарды өткізіп, (тапсырыстар, қателер) кортежін қайтарады."""
    # Шешіміңізді осында жазыңыз
    with open(path, "r", encoding="utf-8") as f:
      lines=[line.strip() for line in f if line.strip()]
    header=lines[0].split(";")
    data_lines=lines[1:]
    orders_dicts=[]
    errors=[]

    for index, line in enumerate(data_lines, start=2):
      parts=line.split(";")
      try:
        parts[6]=int(parts[6])
        parts[7]=int(parts[7])
        d=dict(zip(header,parts))
        d["amount"]=parts[6]*parts[7]
        orders_dicts.append(d)
      except ValueError as error:
        errors.append((index, line))
    return (orders_dicts, errors)
    ...


orders_all, errors = safe_load("orders.txt")
print("Дұрыс тапсырыстар:", len(orders_all))
print("Қате жолдар:", errors)
```

### Тапсырма 26. Санаттар есебі CSV форматында

`csv` модулін қолданып, `orders_all` тізімі бойынша `category_report.csv` файлын жазыңыз. Бағандары: `category`, `orders`, `quantity`, `revenue`. Жолдар түсім бойынша кему ретімен орналасуы керек. Файлды `csv.DictReader` арқылы қайта оқып, `revenue` бағанының қосындысы жалпы түсімге тең екенін тексеріңіз.

**Өзін-өзі тексеру:** Electronics 121500, Books 52500, Stationery 31500.

```python
import csv
cat_stats={}
for o in orders_all:
  category=o["category"]
  if category not in cat_stats:
    cat_stats[category]={"orders":0, "quantity":0, "revenue":0}
  cat_stats[category]["orders"]+=1
  cat_stats[category]["quantity"]+=int(o["quantity"])
  cat_stats[category]["revenue"]+=o["amount"]
sorted_cats=sorted(cat_stats.items(), key=lambda x: x[1]["revenue"], reverse=True)

with open("category_report.csv", "w", encoding="utf-8", newline="") as f:
  writer=csv.DictWriter(f, fieldnames=["category", "orders", "quantity", "revenue"])
  writer.writeheader()
  for category, stats in sorted_cats:
    writer.writerow({"category": category, **stats})
total_revenue_check = 0
with open("category_report.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    print("Файл мазмұны:")
    for row in reader:
        print(f"{row['category']} {row['revenue']}")
        total_revenue_check += int(row["revenue"])

print("\nБарлық түсімдер қосындысы:", total_revenue_check)
# Шешіміңізді осында жазыңыз
```

### Тапсырма 28. Қалалар бойынша жеке файлдар

`os.makedirs()` арқылы `cities` бумасын құрыңыз. Әр қала үшін жеке файл жазыңыз, мысалы `cities/Almaty.txt`. Әр файлда тақырып жолы және сол қаланың тапсырыстары `orders.txt` файлындағы форматта болуы тиіс. Соңында бумадағы барлық файлды оқып, тапсырыс жолдарының жалпы саны дұрыс тапсырыстар санына тең екенін тексеріңіз.

**Өзін-өзі тексеру:** `cities` бумасында 4 файл, оларда барлығы 12 тапсырыс жолы бар.

```python
import os

#  'cities' атты бумаганы құру (егер бұрыннан бар болса, қате бермеуі үшін exist_ok=True қолданамыз)
os.makedirs("cities", exist_ok=True)

# Тапсырыстарды қалалар бойынша топтастыру
city_orders = {}
for o in orders_all:
    city = o["city"]
    if city not in city_orders:
        city_orders[city] = []
    city_orders[city].append(o)

# 3. Әр қала үшін жеке файл жазу (тақырып пен бастапқы форматты сақтай отырып)
for city, orders in city_orders.items():
    file_path = os.path.join("cities", f"{city}.txt")
    with open(file_path, "w", encoding="utf-8") as f:
        # Бастапқы orders.txt файлындағыдай бақырып/тақырыпты жазамыз
        f.write("id;date;customer;city;category;product;price;quantity\n")

        # Әр тапсырысты нүктелі үтірмен біріктіріп жазу
        for o in orders:
            line = f"{o.get('id', o.get('order_id', ''))};{o['date']};{o['customer']};{o['city']};{o['category']};{o['product']};{o['price']};{o['quantity']}\n"
            f.write(line)

# 4. Бумадағы барлық файлдарды оқып, тапсырыс жолдарының жалпы санын тексеру
total_count = 0
for filename in os.listdir("cities"):
    if filename.endswith(".txt"):
        path = os.path.join("cities", filename)
        with open(path, "r", encoding="utf-8") as f:
            lines = [line.strip() for line in f if line.strip()]
            # Тақырып жолын алып тастап, тек тапсырыстарды санау (-1 тақырып үшін)
            order_lines_count = len(lines) - 1
            total_count += order_lines_count
            print(f"{filename}: {order_lines_count} тапсырыс")

print(f"\nБарлық қалалар бойынша жиналған тапсырыстардың жалпы саны: {total_count}")
print(f"orders_all тізіміндегі тапсырыстар санына сәйкес пе?: {total_count == len(orders_all)}")
```

### Тапсырма 29. Қорытынды есеп

Жоғарыда жазылған функцияларды пайдаланып, `report.txt` қорытынды есебін жазатын `build_report(path, report_path)` функциясын жазыңыз. Есепте мыналар болуы керек: талдау мерзімі (ең ерте және ең кеш күн), дұрыс және қате жолдар саны, жалпы түсім, орташа сома, ең көп ақша жұмсаған 3 клиент, санаттардың түсімдегі үлесі және каталогтағы сатылмаған тауарлар. Функцияны шақырып, есепті оқып шығарыңыз.

```python
def build_report(path, report_path):
    """Барлық деректерді жинап, қорытынды есепті report.txt файлына жазады."""
    # 1. safe_load көмегімен қателерді есепке алмай деректерді оқимыз
    orders_list, errors = safe_load(path)

    # 2. Мерзімді анықтау (ең ерте және ең кеш күн)
    dates = [o["date"] for o in orders_list]
    earliest_date = min(dates) if dates else "Жоқ"
    latest_date = max(dates) if dates else "Жоқ"

    # 3. Қаржылық көрсеткіштер
    total_revenue = sum(o["amount"] for o in orders_list)
    avg_amount = total_revenue / len(orders_list) if orders_list else 0

    # 4. Клиенттер статистикасы (customer_stats қолданамыз)
    cstats = customer_stats(orders_list)
    # Ең көп ақша жұмсаған топ-3 клиентті табамыз
    top_spenders = sorted(cstats.items(), key=lambda x: x[1]["total"], reverse=True)[:3]

    # 5. Санаттардың үлесін есептеу
    cat_revenue = {}
    for o in orders_list:
        cat = o["category"]
        cat_revenue[cat] = cat_revenue.get(cat, 0) + o["amount"]

    # 6. Файлға жазу
    with open(report_path, "w", encoding="utf-8") as f:
        f.write("=== ҚОРЫТЫНДЫ ЕСЕП ===\n\n")

        f.write("1. Талдау мерзімі:\n")
        f.write(f"   - Ең ерте күн: {earliest_date}\n")
        f.write(f"   - Ең кеш күн: {latest_date}\n\n")

        f.write("2. Деректер сапасы:\n")
        f.write(f"   - Дұрыс тапсырыстар саны: {len(orders_list)}\n")
        f.write(f"   - Қате жолдар саны: {len(errors)}\n\n")

        f.write("3. Қаржылық көрсеткіштер:\n")
        f.write(f"   - Жалпы түсім: {total_revenue} теңге\n")
        f.write(f"   - Орташа сома: {avg_amount:.1f} теңге\n\n")

        f.write("4. Ең көп ақша жұмсаған топ-3 клиент:\n")
        for name, info in top_spenders:
            f.write(f"   - {name}: {info['total']} теңге ({info['orders']} тапсырыс)\n")
        f.write("\n")

        f.write("5. Санаттардың түсімдегі үлесі:\n")
        for cat, rev in cat_revenue.items():
            percentage = (rev / total_revenue) * 100 if total_revenue > 0 else 0
            f.write(f"   - {cat}: {rev} теңге ({percentage:.1f}%)\n")

# Функцияны орындау
build_report("orders.txt", "report.txt")

# Нәтижелік есепті оқып экранға шығару
with open("report.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

### Тапсырма 30. Нәтижелерді жинау және жүктеп алу

Жұмыс барысында жасалған барлық файлды (`summary.txt`, `category_report.csv`, `customers.json`, `report.txt` және `cities` бумасы) `results` атты жаңа бумаға көшіріңіз. Файлдарды көшіру үшін `shutil.copy()`, ал буманы көшіру үшін `shutil.copytree()` қолданыңыз. Бумадағы файлдарды өлшемімен бірге шығарыңыз. Содан кейін буманы `shutil.make_archive()` арқылы `homework_results.zip` мұрағатына жинап, `files.download()` арқылы компьютерге жүктеп алыңыз.

```python
import shutil
import os

# 1. 'results' атты жаңа бума құрамыз (егер бұрын бар болса, тазалап не өткізіп жіберу үшін)
os.makedirs("results", exist_ok=True)

# 2. Жеке файлдарды results бумасына көшіру
files_to_copy = ["summary.txt", "category_report.csv", "customers.json", "report.txt"]

for file in files_to_copy:
    if os.path.exists(file):
        shutil.copy(file, os.path.join("results", file))

# 3. 'cities' бумасын толығымен results ішіне көшіру
if os.path.exists("cities"):
    dest_cities = os.path.join("results", "cities")
    if os.path.exists(dest_cities):
        shutil.rmtree(dest_cities)  # Егер бұрын болса, өшіріп қайта көшіреміз
    shutil.copytree("cities", dest_cities)

# 4. results бумасындағы файлдарды өлшемімен бірге шығару
print("=== Results бумасындағы файлдар ===")
for root, dirs, files in os.walk("results"):
    for file in files:
        file_path = os.path.join(root, file)
        file_size = os.path.getsize(file_path)
        print(f"{file_path} — {file_size} байт")

# 5. Барлығын homework_results.zip архивіне жинау
shutil.make_archive("homework_results", "zip", "results")
print("\nАрхів сәтті жасалды: homework_results.zip")

# 6. Колледж/платформа ортасында жүктеп алу үшін (Google Colab үшін):
try:
    from google.colab import files
    files.download("homework_results.zip")
    print("Файл жүктеуге жіберілді!")
except ImportError:
    print("Бұл орта Google Colab емес, бірақ архив локальді түрде сақталды.")
```
