### Тапсырма 22. Үшбұрыштың түрі

Үшбұрыштың үш қабырғасының ұзындығы `a`, `b`, `c` берілген. Алдымен мұндай үшбұрыштың болуы мүмкін екенін тексеріңіз: кез келген екі қабырғаның қосындысы үшінші қабырғадан үлкен болуы керек. Үшбұрыш бар болса, оның тең қабырғалы, теңбүйірлі немесе әртүрлі қабырғалы екенін анықтаңыз.

````python
a=int(input())
b=int(input())
c=int(input())
if (a+b)>c and (b+c)>a and (a+c)>b:
  if a==b and a==c:
    print("Тең қабырғалы үшбұрыш")
  elif (a==b and a!=c) or (b==c and b!=a) or (a==c and a!=b):
    print("Теңбүйірлі үшбұрыш")
  elif a!=b and b!=c and a!=c:
    print("Әртүрлі қабырғалы")
else:
  print("Бұндай үшбұрыш жоқ")
````

### Тапсырма 23. Қарапайым калькулятор

`a`, `b` сандары және `operation` айнымалысы (`"+"`, `"-"`, `"*"` немесе `"/"`) берілген. Көрсетілген амалды орындап, нәтижені шығарыңыз. Нөлге бөлуге әрекет жасалса немесе белгісіз амал берілсе, тиісті хабарлама шығарыңыз.

````python
first_number=int(input("Enter the first number: "))
second_number=int(input("Enter the second number: "))
operation=input("Enter the operation: ")
if operation=='+':
  result=first_number+second_number
  print(f"The result is {result}")
elif operation=='-':
  result=first_number-second_number
  print(f"The result is {result}")
elif operation=='*':
  result=first_number*second_number
  print(f"The result is {result}")
elif operation=='/':
  if second_number==0:
    print("Error! The number cannot be divided by zero!")
  else:
    result=first_number/second_number
    print(f"The result is {result}")
else:
  print("Error! Invalid operation!")
````

### Тапсырма 28. Сандарды санау

`numbers = [12, -5, 8, 0, -3, 17, 4, 0, -9]` тізімі берілген. Тізімдегі оң, теріс сандардың және нөлдердің санын бөлек есептеп шығарыңыз.

**Күтілетін нәтиже:**

```
Оң сандар: 4
Теріс сандар: 3
Нөлдер: 2
```

````python
numbers = [12, -5, 8, 0, -3, 17, 4, 0, -9]
pos=0
neg=0
zero=0
for i in range(0, len(numbers)-1):
  if numbers[i]>0:
    pos+=1
  elif numbers[i]<0:
    neg+=1
  else:
    zero+=1
print(f"Оң сандар: {pos}")
print(f"Теріс сандар: {neg}")
print(f"Нөлдер: {zero}")
````

Фибоначчи тізбегінің алғашқы `n` мүшесінен тұратын тізімді қайтаратын `fibonacci(n)` функциясын жазыңыз. Тізбектің алғашқы екі мүшесі 0 және 1, ал әрбір келесі мүше алдыңғы екі мүшенің қосындысына тең.

### Тапсырма 44. Тізімнің статистикасы

Сандар тізімін қабылдап, оның ең кіші мәнін, ең үлкен мәнін және орташа мәнін кортеж түрінде қайтаратын `stats(numbers)` функциясын жазыңыз. `min()` және `max()` функцияларын қолданбаңыз. Бос тізім берілсе, функция `None` қайтаруы керек.

**Күтілетін нәтиже:**

```
stats([4, 9, 1, 7, 4]) -> (1, 9, 5.0)
stats([]) -> None
```

````python
def stats(numbers):
    if not numbers:
        return None
    minimum = numbers[0]
    maximum = numbers[0]
    total = 0
    for num in numbers:
        if num < minimum:
            minimum = num
        if num > maximum:
            maximum = num
        total += num
    average = total / len(numbers)
    return (minimum, maximum, average)
print(f"stats([4, 9, 1, 7, 4]) -> {stats([4, 9, 1, 7, 4])}")
print(f"stats([]) -> {stats([])}")
````

Функция құпиясөз сенімді болса "Құпиясөз сенімді" деп, әйтпесе қай талап орындалмағанын көрсететін хабарламаны қайтаруы тиіс. Жол әдістерінен `isdigit()` және `isupper()` пайдалы болуы мүмкін.

**Күтілетін нәтиже:**

```
validate_password("abc") -> Құпиясөз кемінде 8 символдан тұруы керек
```

`operations = [("deposit", 5000), ("withdraw", 3000), ("withdraw", 15000), ("deposit", 2000), ("withdraw", 4000)]`

Әр операцияны орындайтын `process(balance, operation)` функциясын жазыңыз. `deposit` балансты көбейтеді, `withdraw` азайтады. Қаражат жеткіліксіз болса, операция орындалмай, ескерту шығуы керек. Барлық операциядан кейінгі соңғы балансты шығарыңыз.

**Күтілетін нәтиже:**

```
Соңғы баланс: 10000
```

````python
def process(balance, operation):
    op_type, amount = operation
    if op_type == "deposit":
        return balance + amount
    elif op_type == "withdraw":
      if balance>=amount:
        return balance - amount
    return balance

balance = 10000
operations = [("deposit", 5000), ("withdraw", 3000), ("withdraw", 15000), ("deposit", 2000), ("withdraw", 4000)]

for op in operations:
    balance = process(balance, op)

print(f"Соңғы баланс: {balance}")
````

### Тапсырма 48. Студенттер журналы

Студенттердің баллдары сөздікте берілген:

`journal = {"Aigerim": [85, 92, 78], "Nurlan": [65, 70, 58], "Dana": [95, 98, 91], "Arman": [45, 52, 40]}`

Келесі функцияларды жазыңыз: орташа баллды есептейтін `average(scores)`, баллға сәйкес бағаны (A, B, C, F) қайтаратын `get_grade(score)`, үздік студенттің атын қайтаратын `best_student(journal)`. Осы функцияларды қолданып, әр студенттің аты, орташа баллы мен бағасы көрсетілген кестені және үздік студентті шығарыңыз.

````python
journal = {
    "Aigerim": [85, 92, 78],
    "Nurlan": [65, 70, 58],
    "Dana": [95, 98, 91],
    "Arman": [45, 52, 40],
}

def average(scores):
  return round((sum(scores) / len(scores)),1)

def get_grade(score):
  if score >= 90:
    return "A"
  elif score >= 75:
    return "B"
  elif score >= 60:
    return "C"
  else:
    return "F"
def best_student(journal_dict):
  best_name = None
  highest_avg = -1
  for name, scores in journal_dict.items():
    avg = average(scores)
    if avg > highest_avg:
      highest_avg = avg
      best_name = name
  return best_name

print(f"Аты | Орташа балл | Бағасы")
for name, scores in journal.items():
  avg = average(scores)
  grade = get_grade(avg)
  print(f"{name} | {avg} | {grade}")

print(f"Үздік студент: {best_student(journal)}")
````

Сан табылғанда цикл тоқтап, неше әрекетте табылғаны көрсетілуі керек. Тексеру логикасын `check_guess(secret, guess)` функциясына бөліп жазыңыз.

**Күтілетін нәтиже:**

```
50: Ойлаған сан кішірек
25: Ойлаған сан үлкенірек
37: Ойлаған сан үлкенірек
45: Ойлаған сан кішірек
42: Таптыңыз!
Әрекеттер саны: 5
```

### Тапсырма 50. Цезарь шифры

Мәтінді Цезарь шифрымен шифрлайтын `encrypt(text, shift)` функциясын жазыңыз: әрбір латын әрпі алфавит бойынша `shift` орынға жылжытылады, ал алфавит соңына жеткенде басынан жалғасады. Бас және кіші әріптердің регистрі сақталуы, ал әріп емес символдар өзгеріссіз қалуы керек. Содан кейін шифрланған мәтінді бастапқы күйіне қайтаратын `decrypt(text, shift)` функциясын жазыңыз. `ord()` және `chr()` функциялары пайдалы болуы мүмкін.

**Күтілетін нәтиже:**

```
encrypt("Hello, World!", 3) -> Khoor, Zruog!
decrypt("Khoor, Zruog!", 3) -> Hello, World!
```

````python
def encrypt(text, shift):
    result = ""
    for char in text:
        if 'A' <= char <= 'Z':
            code = (ord(char) - ord('A') + shift) % 26 + ord('A')
            result += chr(code)
        elif 'a' <= char <= 'z':
            code = (ord(char) - ord('a') + shift) % 26 + ord('a')
            result += chr(code)
        else:
            result += char
    return result

def decrypt(text, shift):
    return encrypt(text, -shift)
print(f'encrypt("Hello, World!", 3) -> {encrypt("Hello, World!", 3)}')
print(f'decrypt("Khoor, Zruog!", 3) -> {decrypt("Khoor, Zruog!", 3)}')
````
