5-тапсырма

```python
count=df["Churn"].value_counts() #әр мәннің саны
percent=(df["Churn"].value_counts(normalize=True)*100).round(2)
table=pd.DataFrame({

    "count": count,
    "percent %": percent
})
print(table)
print()
print("Бірегей мәндер саны:", df["Churn"].nunique())
print()
display(df.nlargest(5, "tenure")[["customerID","tenure", "MonthlyCharges", "TotalCharges"]])
```

6-тапсырма, жаңа екі баған

```python
df["tenure_year"]=(df["tenure"]/12).round(1) #айды жылға айналдыру
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[0, 32, 72],
    labels=["ортадан төмен", "ортадан жоғары"]
)
display(df[["tenure", "tenure_year", "tenure_group"]])
```

7-тапсырма, фильтрлеу

```python
service=df[df["InternetService"]=="Fiber optic"][["customerID", "InternetService"]]
security=df[(df["OnlineSecurity"]=="No") & (df["OnlineBackup"]=="No")][["customerID", "OnlineSecurity", "OnlineBackup"]]
churn_female=df.query("gender=='Female' and Churn=='Yes'")[["customerID", "gender", "Churn"]]
churn_male=df.query("gender=='Male' and Churn=='Yes'")[["customerID", "gender", "Churn"]]
print("Fiber optic клиенттер саны:", len(service))
print("OnlineSecurity='No' және OnlineBackup='No' саны:", len(security))
print("Кеткен әйелдер саны:", len(churn_female))
print("Кеткен ерлер саны:", len(churn_male))
display(service)
display(security)
display(churn_female)
display(churn_male)
```

8-тапсырма, топтау

```python
grouped=df.groupby("gender").agg(

        tenure_mean=("tenure", "mean"),
        MonthlyCharges_min=("MonthlyCharges", "min"),
        MonthlyCharges_max=("MonthlyCharges", "max"),
        client_count=("Churn", "count")

).sort_values("tenure_mean", ascending=False)
display(grouped)
```

9-тапсырма, жиынтық кесте

```python
pivot=df.pivot_table(
    values="MonthlyCharges",
    index="Contract",
    columns="InternetService",
    aggfunc="mean").round(2)
display(pivot)
```

Ең жоғары орташа айлық төлем (104.57) Two year + Fiber optic тобында байқалады — бұл ұзақ мерзімді келісімшарттағы жоғары жылдамдықты интернет пен қосымша қызметтердің жиынтық құнына байланысты.Ең төменгі орташа айлық төлем (20.41) Month-to-month + No (интернетсіз) тобында тіркелді — бұл қызметтер жиынтығының ең қарапайым әрі арзан болуын білдіреді.

10-тапсырма, нәтижелерді сақтау

```python
# Тазаланған кестені сақтау
df.to_csv("telco_churn_cleaned.csv", index=False)

# Негізгі жиынтық кестелерді сақтау
grouped.to_csv("summary_by_gender.csv")
pivot.to_csv("pivot_contract_internet_charges.csv")

print("Файлдар сәтті сақталды.")
files.download("telco_churn_cleaned.csv")
files.download("summary_by_gender.csv")
files.download("pivot_contract_internet_charges.csv")
```

Қорытынды

Telco Customer Churn датасетін талдау нәтижесінде клиенттердің 26.54%-ы (1869 клиент) қызметтен кеткенін, ал 73.46%-ы (5174 клиент) қалғанын көрдік. Жыныс бойынша топтауда ер (3555) және әйел (3488) клиенттердің орташа қызмет мерзімі (32.50 және 32.24 ай) мен айлық төлем ауқымы (18.25–118.35 және 18.40–118.75) бір-біріне өте жақын болды. Келісімшарт түрі мен интернет қызметі бойынша жиынтық кестеде орташа айлық төлемнің ең жоғары мәні 104.57 (Two year + Fiber optic), ал ең төменгі мәні 20.41 (Month-to-month + No) шықты. Деректердің шектеулері ретінде TotalCharges бағанындағы бастапқы бос/мәтін мәндердің tenure=0 жағдайларымен байланыстылығын және талдаудың бір уақыттық тарихи іріктемеге негізделгенін атап өтуге болады. Келесі қадамда кету (Churn) көрсеткішін келісімшарт түрі (Contract), төлем әдісі (PaymentMethod) және қызмет мерзімі (tenure_group) бойынша қосымша салыстыруға болады.
